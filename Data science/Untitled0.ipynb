```python
#program 1: Evaluating of (x^2-4)/(x-2) as x approaches 2
import numpy as np
#Define the function
def f(x):
  return (x**2-4)/(x-2)

#print where we are evaluating the limit
a=2.0
#step size (approaching from left to right)
h_value=[0.1,0.01,0.001,0.00001]

print("Evaluating limit of f(x)=(x^2)/(x-2) as  x approaches to 2")
print("="*60)

print(f"{'h':>10} {'x(left':15}{'f(x) left':>15}{'x(right)':>15}")
print("-"*60)
for h in h_value:
  x_left=a-h    # Approach from the
  x_right=a+h   #Approcah from right
  f_left=f(x_left)
  f_right=f(x_right)
  print(f"{h:>10}{x_left:>15.5f}{f_left:>15.5f}{x_right:>15.5f}{x_right:>15.5f}")

print("="*60)
print("As h--> 0 both left and right limits approaches to 4")
print("Therefore , limit exist and equal to 4")
```
