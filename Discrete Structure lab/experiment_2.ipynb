```python
A = {1, 2, 3}

R = {(1, 1), (2, 2), (3, 3),
     (1, 2), (2, 1)}

reflexive = all((a, a) in R for a in A)

symmetric = all((b, a) in R for (a, b) in R)

transitive = all(
    (a, c) in R
    for (a, b) in R
    for (b, c) in R
)

print("Reflexive:", reflexive)
print("Symmetric:", symmetric)
print("Transitive:", transitive)
```
