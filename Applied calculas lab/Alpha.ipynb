```python
import numpy as np

def f(x):
    return (x**2 - 1)/(x - 1)

a = 1  # point of interest

# Values approaching from left
left_values = a - np.logspace(-1, -6, 6)

# Values approaching from right
right_values = a + np.logspace(-1, -6, 6)

print("Left-hand values:")
for x in left_values:
    print(x, f(x))

print("\nRight-hand values:")
for x in right_values:
    print(x, f(x))
```

```python
import numpy as np
import matplotlib.pyplot as plt

plt.figure(figsize=(12,8))

# -------------------------------
# 1) Limit behavior near a point
# f(x) = (x^2 - 1)/(x - 1), limit at x = 1
# -------------------------------
x1 = np.linspace(0.5, 1.5, 400)
y1 = (x1**2 - 1)/(x1 - 1)

plt.subplot(2,2,1)
plt.plot(x1, y1, label='f(x)')
plt.scatter(1, 2)  # limit point
plt.axvline(1, linestyle='--')
plt.title("Limit Behavior near x=1")
plt.legend()


# -------------------------------
# 2) Removable Discontinuity
# Same function with a hole at x=1
# -------------------------------
plt.subplot(2,2,2)
plt.plot(x1, y1)
plt.scatter(1, 2)  # hole shown
plt.title("Removable Discontinuity")


# -------------------------------
# 3) Jump Discontinuity
# f(x) = 1 for x<0, 2 for x>=0
# -------------------------------
x_left = np.linspace(-2, 0, 200)
x_right = np.linspace(0, 2, 200)

plt.subplot(2,2,3)
plt.plot(x_left, [1]*len(x_left))
plt.plot(x_right, [2]*len(x_right))
plt.axvline(0, linestyle='--')
plt.title("Jump Discontinuity")


# -------------------------------
# 4) Infinite Discontinuity
# f(x) = 1/x
# -------------------------------
x2_left = np.linspace(-2, -0.1, 200)
x2_right = np.linspace(0.1, 2, 200)

plt.subplot(2,2,4)
plt.plot(x2_left, 1/x2_left)
plt.plot(x2_right, 1/x2_right)
plt.axvline(0, linestyle='--')
plt.title("Infinite Discontinuity")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Function and its derivative
def f(x):
    return x**3 - 3*x**2 + 2

def df(x):
    return 3*x**2 - 6*x

x = np.linspace(-1, 4, 400)

# Plot both on same graph
plt.figure(figsize=(8,6))
plt.plot(x, f(x), label="f(x) = x^3 - 3x^2 + 2")
plt.plot(x, df(x), label="f'(x) = 3x^2 - 6x")

# Critical points where derivative = 0
critical_x = [0, 2]
critical_y = [f(0), f(2)]

plt.scatter(critical_x, critical_y)

plt.axhline(0, linestyle="--")
plt.axvline(0, linestyle="--")

plt.title("Function and its Derivative")
plt.legend()
plt.show()

# Analysis (printed insights)
print("Critical points at x = 0 and x = 2")
print("Check derivative sign:")

test_points = [-0.5, 1, 3]
for t in test_points:
    print(f"f'({t}) =", df(t))
```

```python
import sympy as sp
import numpy as np

# Symbol
x = sp.symbols('x')

# Choose function
f = sp.exp(x)   # you can change to sp.sin(x), sp.cos(x), etc.

# Degree of expansion
degree = 5

# Maclaurin series (around x = 0)
maclaurin = sp.series(f, x, 0, degree+1).removeO()

# Taylor series (around x = 1)
taylor = sp.series(f, x, 1, degree+1).removeO()

print("Function:", f)
print("\nMaclaurin series (degree", degree, "):")
print(maclaurin)

print("\nTaylor series at x=1 (degree", degree, "):")
print(taylor)

# Convert to numerical functions
f_num = sp.lambdify(x, f)
mac_num = sp.lambdify(x, maclaurin)

```

```python
# Test points
points = [0.2, 0.5, 1.0, 1.5]

print("\nComparison of actual value vs approximation:\n")
for p in points:
    actual = f_num(p)
    approx = mac_num(p)
    error = abs(actual - approx)
    print(f"x = {p}")
    print("Actual value      =", actual)
    print("Maclaurin approx  =", approx)
    print("Error             =", error)
    print()
```
