# Experiment 7: Lagrange Multipliers for Constrained Optimization

**Objective:** To solve constrained optimization problems using the method of Lagrange multipliers.

---

## What is the Lagrange Multiplier Method?

We want to find the **maximum or minimum** of a function **f(x, y)**  
subject to a **constraint g(x, y) = c**

The key idea is:

At the optimal point, the gradients of f and g must be parallel:

**grad f = lambda * grad g**

This gives us the system of equations:
- df/dx = lambda * dg/dx
- df/dy = lambda * dg/dy
- g(x, y) = c

We solve these equations to find the critical points.


## Step 1: Install and Import Required Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from sympy import symbols, solve, diff, lambdify, latex

print("Libraries imported successfully!")
```

---
## Step 2: Define the Core Function

This function takes f(x,y), g(x,y), and the constant c,  
then finds and plots the critical points.

```python
def lagrange_optimizer(f_expr, g_expr, c_val, x_range=(-5, 5), y_range=(-5, 5)):
    """
    Finds extrema of f(x,y) subject to g(x,y) = c using Lagrange Multipliers.

    Parameters:
        f_expr  : objective function in terms of x, y (sympy expression)
        g_expr  : constraint function in terms of x, y (sympy expression)
        c_val   : the constant value c in g(x,y) = c
        x_range : tuple for x-axis plot range
        y_range : tuple for y-axis plot range
    """

    x, y, lam = symbols('x y lambda')

    # Partial derivatives of f
    fx = diff(f_expr, x)
    fy = diff(f_expr, y)

    # Partial derivatives of g
    gx = diff(g_expr, x)
    gy = diff(g_expr, y)

    # Lagrange conditions
    eq1 = fx - lam * gx          # df/dx = lambda * dg/dx
    eq2 = fy - lam * gy          # df/dy = lambda * dg/dy
    eq3 = g_expr - c_val         # g(x,y) = c

    print("Lagrange Equations to Solve:")
    print(f"  Equation 1 (df/dx = lam * dg/dx): {eq1} = 0")
    print(f"  Equation 2 (df/dy = lam * dg/dy): {eq2} = 0")
    print(f"  Equation 3 (constraint):           {eq3} = 0")
    print()

    # Solve the system
    solutions = solve([eq1, eq2, eq3], [x, y, lam], dict=True)

    print("Critical Points Found:")
    real_points = []

    for sol in solutions:
        xv = sol[x]
        yv = sol[y]
        # Keep only real-valued solutions
        if xv.is_real and yv.is_real:
            fv = f_expr.subs([(x, xv), (y, yv)])
            print(f"  x = {float(xv):.4f}, y = {float(yv):.4f}, f(x,y) = {float(fv):.4f}")
            real_points.append((float(xv), float(yv), float(fv)))

    if not real_points:
        print("  No real critical points found.")
        return

    # Identify max and min among critical points
    f_values = [pt[2] for pt in real_points]
    max_val = max(f_values)
    min_val = min(f_values)
    print(f"\n  Maximum value of f = {max_val:.4f}")
    print(f"  Minimum value of f = {min_val:.4f}")

    # ---- Plotting ----
    f_num = lambdify((x, y), f_expr, 'numpy')
    g_num = lambdify((x, y), g_expr, 'numpy')

    X = np.linspace(x_range[0], x_range[1], 400)
    Y = np.linspace(y_range[0], y_range[1], 400)
    X, Y = np.meshgrid(X, Y)
    F = f_num(X, Y)
    G = g_num(X, Y)

    plt.figure(figsize=(8, 6))

    # Contour plot of f
    contour = plt.contourf(X, Y, F, levels=30, cmap='viridis', alpha=0.6)
    plt.colorbar(contour, label='f(x, y)')

    # Constraint curve g(x,y) = c
    plt.contour(X, Y, G, levels=[c_val], colors='red', linewidths=2)

    # Plot critical points
    for pt in real_points:
        color = 'yellow' if pt[2] == max_val else 'cyan'
        label = f'Max f={pt[2]:.2f}' if pt[2] == max_val else f'Min f={pt[2]:.2f}'
        plt.plot(pt[0], pt[1], 'o', color=color, markersize=12, label=label)
        plt.annotate(f'({pt[0]:.2f}, {pt[1]:.2f})', xy=(pt[0], pt[1]),
                     xytext=(pt[0]+0.2, pt[1]+0.2), fontsize=9, color='white')

    plt.title('Lagrange Multiplier: Constraint Curve and Critical Points')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.legend(loc='upper right')
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```

---
## Step 3: Worked Example

**Problem:**  
Find the maximum and minimum values of  
**f(x, y) = x^2 + y^2**  
subject to the constraint  
**g(x, y) = x + y = 1**

**Interpretation:** We are finding the point on the line x + y = 1 that is closest and farthest from the origin.

```python
x, y = symbols('x y')

# Define objective function and constraint
f = x**2 + y**2
g = x + y
c = 1

print("Example:")
print(f"  Objective  : f(x, y) = x^2 + y^2")
print(f"  Constraint : g(x, y) = x + y = {c}")
print()

lagrange_optimizer(f, g, c, x_range=(-2, 2), y_range=(-2, 2))
```

**Expected Result:**  
The minimum of f = 0.5 is at the point (0.5, 0.5).  
This is the point on the line x + y = 1 closest to the origin.
