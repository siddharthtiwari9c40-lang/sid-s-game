```python
# ============================================
# FOUNDATIONAL PYTHON REVIEW
# ============================================

# 1. FUNCTIONS
def square(x):
    """Function to square a number"""
    return x * x

result = square(5)
print("Square of 5:", result)

# 2. FACTORIAL (Important for Taylor Series)
def factorial(n):
    """Calculate factorial of n"""
    if n == 0 or n == 1:
        return 1
    result = 1
    for i in range(2, n + 1):
        result = result * i
    return result

print("5! =", factorial(5))

# 3. POWER FUNCTION
def power(base, exponent):
    """Calculate base raised to exponent"""
    result = 1
    for i in range(exponent):
        result = result * base
    return result

print("2^5 =", power(2, 5))

# 4. ABSOLUTE VALUE (for error calculation)
def absolute(x):
    """Return absolute value"""
    if x < 0:
        return -x
    return x

print("|-5| =", absolute(-5))

# 5. LISTS (to store series terms)
terms = []
for i in range(5):
    terms.append(i * i)
print("List of squares:", terms)
```

```python
# ============================================
# EXAMPLE 1: Maclaurin Series for e^x
# ============================================

import math

def factorial(n):
    """Calculate factorial of n"""
    if n == 0 or n == 1:
        return 1
    result = 1
    for i in range(2, n + 1):
        result = result * i
    return result

def exponential_maclaurin(x, n_terms):
    """
    Compute e^x using Maclaurin series
    e^x = 1 + x + x^2/2! + x^3/3! + ... + x^n/n!

    Parameters:
    x - point at which to evaluate
    n_terms - number of terms in the series
    """
    result = 0.0

    print("\nComputing e^{:.2f} using {} terms:".format(x, n_terms))
    print("-" * 50)

    for n in range(n_terms):
        term = (x ** n) / factorial(n)
        result += term
        print("Term {}: x^{}/{}! = {:.6f}, Sum so far = {:.6f}".format(
            n, n, n, term, result))

    return result

# Test the function
x_value = 1.0
num_terms = 20

approximation = exponential_maclaurin(x_value, num_terms)
actual_value = math.exp(x_value)
error = abs(actual_value - approximation)

print("\n" + "=" * 50)
print("RESULTS:")
print("Approximation: {:.10f}".format(approximation))
print("Actual e^{}: {:.10f}".format(x_value, actual_value))
print("Absolute Error: {:.10f}".format(error))
print("Relative Error: {:.10f}%".format((error/actual_value) * 100))
```

```python
# ============================================
# EXAMPLE 2: Maclaurin Series for sin(x)
# ============================================

import math

def factorial(n):
    """Calculate factorial of n"""
    if n == 0 or n == 1:
        return 1
    result = 1
    for i in range(2, n + 1):
        result = result * i
    return result

def sine_maclaurin(x, n_terms):
    """
    Compute sin(x) using Maclaurin series
    sin(x) = x - x^3/3! + x^5/5! - x^7/7! + ...

    Only odd powers with alternating signs
    """
    result = 0.0

    print("\nComputing sin({:.8f}) using {} terms:".format(x, n_terms))
    print("-" * 50)

    for n in range(n_terms):
        power = 2 * n + 1  # Only odd powers: 1, 3, 5, 7, ...
        sign = (-1) ** n   # Alternating signs: +, -, +, -, ...

        term = sign * (x ** power) / factorial(power)
        result += term

        print("Term {}: {}*x^{}/{}! = {:.6f}, Sum = {:.6f}".format(
            n, "+" if sign > 0 else "-", power, power, term, result))

    return result

# Test the function
x_value = math.pi / 6  # 45 degrees in radians
num_terms = 8

approximation = sine_maclaurin(x_value, num_terms)
actual_value = math.sin(x_value)
error = abs(actual_value - approximation)

print("\n" + "=" * 50)
print("RESULTS:")
print("x = {:.6f} radians ({:.2f} degrees)".format(x_value, math.degrees(x_value)))
print("Approximation: {:.10f}".format(approximation))
print("Actual sin(x): {:.10f}".format(actual_value))
print("Absolute Error: {:.10e}".format(error))
```

```python
# ============================================
# EXAMPLE 3: Maclaurin Series for cos(x)
# ============================================

import math

def factorial(n):
    if n == 0 or n == 1:
        return 1
    result = 1
    for i in range(2, n + 1):
        result = result * i
    return result

def cosine_maclaurin(x, n_terms):
    """
    Compute cos(x) using Maclaurin series
    cos(x) = 1 - x^2/2! + x^4/4! - x^6/6! + ...

    Only even powers with alternating signs
    """
    result = 0.0

    print("\nComputing cos({:.5f}) using {} terms:".format(x, n_terms))
    print("-" * 50)

    for n in range(n_terms):
        power = 2 * n      # Only even powers: 0, 2, 4, 6, ...
        sign = (-1) ** n   # Alternating signs: +, -, +, -, ...

        term = sign * (x ** power) / factorial(power)
        result += term

        print("Term {}: {}*x^{}/{}! = {:.6f}, Sum = {:.6f}".format(
            n, "+" if sign > 0 else "-", power, power, term, result))

    return result

# Test
x_value = math.pi / 3  # 60 degrees
num_terms = 8

approximation = cosine_maclaurin(x_value, num_terms)
actual_value = math.cos(x_value)
error = abs(actual_value - approximation)

print("\n" + "=" * 50)
print("RESULTS:")
print("Approximation: {:.10f}".format(approximation))
print("Actual cos(x): {:.10f}".format(actual_value))
print("Absolute Error: {:.10e}".format(error))
```

```python
# ============================================
# EXAMPLE 4: Derivatives Using Function Definitions
# ============================================

import math

def factorial(n):
    if n == 0 or n == 1:
        return 1
    result = 1
    for i in range(2, n + 1):
        result = result * i
    return result

# Define function and its derivatives
def f(x):
    """Original function: e^x"""
    return math.exp(x)

def f_derivative_1(x):
    """First derivative: e^x"""
    return math.exp(x)

def f_derivative_2(x):
    """Second derivative: e^x"""
    return math.exp(x)

def f_derivative_3(x):
    """Third derivative: e^x"""
    return math.exp(x)

def taylor_series_exponential(x, a, n_terms):
    """
    Compute Taylor series of e^x around point a
    f(x) = f(a) + f'(a)(x-a) + f''(a)(x-a)^2/2! + ...
    """
    # Store derivative functions in a list
    derivatives = [f, f_derivative_1, f_derivative_2, f_derivative_3]

    result = 0.0
    h = x - a  # (x - a)

    print("\nTaylor series of e^x around a = {:.2f}".format(a))
    print("Evaluating at x = {:.2f}".format(x))
    print("-" * 60)

    for n in range(min(n_terms, len(derivatives))):
        # Get the nth derivative value at point a
        derivative_at_a = derivatives[n](a)

        # Calculate the term
        term = derivative_at_a * (h ** n) / factorial(n)
        result += term

        print("Term {}: f^({})({:.1f}) * ({:.2f})^{} / {}! = {:.6f}".format(
            n, n, a, h, n, n, term))
        print("        Sum so far = {:.6f}".format(result))

    return result

# Test
x_value = 2.0
a_value = 1.0
num_terms = 5

approximation = taylor_series_exponential(x_value, a_value, num_terms)
actual_value = math.exp(x_value)

print("\n" + "=" * 60)
print("RESULTS:")
print("Taylor approximation: {:.10f}".format(approximation))
print("Actual e^{}: {:.10f}".format(x_value, actual_value))
print("Error: {:.10e}".format(abs(actual_value - approximation)))
```

```python
# ============================================
# EXAMPLE: Error Analysis vs Degree
# ============================================

import math

def factorial(n):
    if n == 0 or n == 1:
        return 1
    result = 1
    for i in range(2, n + 1):
        result = result * i
    return result

def exponential_maclaurin(x, n_terms):
    """Compute e^x using Maclaurin series"""
    result = 0.0
    for n in range(n_terms):
        result += (x ** n) / factorial(n)
    return result

def analyze_error_vs_degree(x, max_degree):
    """
    Analyze how error decreases as we increase polynomial degree
    """
    actual_value = math.exp(x)

    print("\nError Analysis for e^{:.2f}".format(x))
    print("Actual value: {:.10f}".format(actual_value))
    print("\n" + "=" * 70)
    print("{:^10} | {:^20} | {:^15} | {:^15}".format(
        "Degree", "Approximation", "Abs Error", "Rel Error %"))
    print("=" * 70)

    for degree in range(1, max_degree + 1):
        approx = exponential_maclaurin(x, degree)
        abs_error = abs(actual_value - approx)
        rel_error = (abs_error / actual_value) * 100

        print("{:^10} | {:^20.10f} | {:^15.10e} | {:^15.10f}".format(
            degree, approx, abs_error, rel_error))

# Test for different x values
print("Case 1: Small x value")
analyze_error_vs_degree(0.5, 10)

print("\n\nCase 2: Medium x value")
analyze_error_vs_degree(2.0, 15)

print("\n\nCase 3: Larger x value")
analyze_error_vs_degree(5.0, 20)
```

```python
# ============================================
# COMPREHENSIVE TAYLOR SERIES CALCULATOR
# ============================================

import math

def factorial(n):
    """Calculate factorial"""
    if n == 0 or n == 1:
        return 1
    result = 1
    for i in range(2, n + 1):
        result = result * i
    return result

def maclaurin_exponential(x, n_terms):
    """e^x Maclaurin series"""
    result = 0.0
    for n in range(n_terms):
        result += (x ** n) / factorial(n)
    return result

def maclaurin_sine(x, n_terms):
    """sin(x) Maclaurin series"""
    result = 0.0
    for n in range(n_terms):
        power = 2 * n + 1
        sign = (-1) ** n
        result += sign * (x ** power) / factorial(power)
    return result

def maclaurin_cosine(x, n_terms):
    """cos(x) Maclaurin series"""
    result = 0.0
    for n in range(n_terms):
        power = 2 * n
        sign = (-1) ** n
        result += sign * (x ** power) / factorial(power)
    return result

def numerical_derivative(func, x, order, h=0.0001):
    """Compute numerical derivative"""
    if order == 0:
        return func(x)
    elif order == 1:
        return (func(x + h) - func(x - h)) / (2 * h)
    elif order == 2:
        return (func(x + h) - 2*func(x) + func(x - h)) / (h * h)
    else:
        def deriv_func(t):
            return numerical_derivative(func, t, order - 1, h)
        return numerical_derivative(deriv_func, x, 1, h)

def taylor_series_general(func, x, a, n_terms):
    """General Taylor series using numerical derivatives"""
    result = 0.0
    h = x - a

    for n in range(n_terms):
        derivative_at_a = numerical_derivative(func, a, n)
        term = derivative_at_a * (h ** n) / factorial(n)
        result += term

    return result

def print_menu():
    """Display menu"""
    print("\n" + "=" * 60)
    print(" " * 15 + "TAYLOR SERIES CALCULATOR")
    print("=" * 60)
    print("1. Compute e^x (Maclaurin)")
    print("2. Compute sin(x) (Maclaurin)")
    print("3. Compute cos(x) (Maclaurin)")
    print("4. Error analysis for a function")
    print("5. Custom function (numerical derivatives)")
    print("6. Exit")
    print("=" * 60)

def main():
    """Main program"""

    while True:
        print_menu()
        choice = input("\nEnter your choice (1-6): ")

        if choice == '1':
            x = float(input("Enter x value: "))
            n = int(input("Enter number of terms: "))

            approx = maclaurin_exponential(x, n)
            actual = math.exp(x)
            error = abs(actual - approx)

            print("\n" + "-" * 60)
            print("e^{:.2f}:".format(x))
            print("  Approximation ({} terms): {:.10f}".format(n, approx))
            print("  Actual value:            {:.10f}".format(actual))
            print("  Absolute error:          {:.10e}".format(error))
            print("  Relative error:          {:.6f}%".format((error/actual)*100))

        elif choice == '2':
            x = float(input("Enter x value (in radians): "))
            n = int(input("Enter number of terms: "))

            approx = maclaurin_sine(x, n)
            actual = math.sin(x)
            error = abs(actual - approx)

            print("\n" + "-" * 60)
            print("sin({:.4f}) = sin({:.2f}°):".format(x, math.degrees(x)))
            print("  Approximation ({} terms): {:.10f}".format(n, approx))
            print("  Actual value:            {:.10f}".format(actual))
            print("  Absolute error:          {:.10e}".format(error))

        elif choice == '3':
            x = float(input("Enter x value (in radians): "))
            n = int(input("Enter number of terms: "))

            approx = maclaurin_cosine(x, n)
            actual = math.cos(x)
            error = abs(actual - approx)

            print("\n" + "-" * 60)
            print("cos({:.4f}) = cos({:.2f}°):".format(x, math.degrees(x)))
            print("  Approximation ({} terms): {:.10f}".format(n, approx))
            print("  Actual value:            {:.10f}".format(actual))
            print("  Absolute error:          {:.10e}".format(error))

        elif choice == '4':
            print("\nError Analysis for sin(x)")
            x = float(input("Enter x value: "))
            max_terms = int(input("Enter maximum terms to test: "))

            actual = math.sin(x)
            print("\n{:^10} | {:^18} | {:^15}".format("Terms", "Approximation", "Error"))
            print("-" * 50)

            for n in range(1, max_terms + 1):
                approx = maclaurin_sine(x, n)
                error = abs(actual - approx)
                print("{:^10} | {:^18.10f} | {:^15.6e}".format(n, approx, error))

        elif choice == '5':
            print("\nCustom Function: e^(x^2)")
            x = float(input("Enter x value: "))
            a = float(input("Enter center point a: "))
            n = int(input("Enter number of terms: "))

            def custom_func(t):
                return math.exp(t * t)

            approx = taylor_series_general(custom_func, x, a, n)
            actual = custom_func(x)
            error = abs(actual - approx)

            print("\n" + "-" * 60)
            print("Taylor series around a = {:.2f}:".format(a))
            print("  Approximation at x={:.2f}: {:.10f}".format(x, approx))
            print("  Actual value:             {:.10f}".format(actual))
            print("  Error:                    {:.10e}".format(error))

        elif choice == '6':
            print("\nThank you for using Taylor Series Calculator!")
            break

        else:
            print("\nInvalid choice! Please try again.")

        input("\nPress Enter to continue...")

# Run the program
if __name__ == "__main__":
    main()
```
