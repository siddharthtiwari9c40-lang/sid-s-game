```python
import numpy as np

def numerical_derivative(f, x, h=1e-5):
    """
    Calculate derivative of f at point x using finite difference method
    f: function
    x: point at which to calculate derivative
    h: small step size
    """
    return (f(x + h) - f(x - h)) / (2 * h)

# Example usage
def f(x):
    return x**2

x_point = 3
derivative = numerical_derivative(f, x_point)
print(f"Derivative of f(x) = x^2 at x = {x_point} is: {derivative}")
print(f"Analytical derivative: {2 * x_point}")
```

```python
import numpy as np

def forward_derivative(f, x, h=1e-5):
    return (f(x + h) - f(x)) / h

def backward_derivative(f, x, h=1e-5):
    return (f(x) - f(x - h)) / h

def central_derivative(f, x, h=1e-5):
    return (f(x + h) - f(x - h)) / (2 * h)

def f(x):
    return 1 / (x**2 + x)

x_point = 5

forward = forward_derivative(f, x_point)
backward = backward_derivative(f, x_point)
central = central_derivative(f, x_point)



print(f"Forward Derivative at x = {x_point}: {forward}")
print(f"Backward Derivative at x = {x_point}: {backward}")
print(f"Central Derivative at x = {x_point}: {central}")
```

```python
import numpy as np

def partial_derivative_x(f, x, y, h=1e-5):
    """
    Calculate partial derivative with respect to x
    """
    return (f(x + h, y) - f(x - h, y)) / (2 * h)

def partial_derivative_y(f, x, y, h=1e-5):
    """
    Calculate partial derivative with respect to y
    """
    return (f(x, y + h) - f(x, y - h)) / (2 * h)

# Example usage
def f(x, y):
    return x**2 + y**2

x_point, y_point = 2, 3
df_dx = partial_derivative_x(f, x_point, y_point)
df_dy = partial_derivative_y(f, x_point, y_point)

print(f"At point ({x_point}, {y_point}):")
print(f"∂f/∂x = {df_dx}")
print(f"∂f/∂y = {df_dy}")
```

```python
import numpy as np

def gradient_2d(f, x, y, h=1e-5):
    """
    Calculate gradient vector [∂f/∂x, ∂f/∂y] at point (x, y)
    Returns: numpy array
    """
    df_dx = (f(x + h, y) - f(x - h, y)) / (2 * h)
    df_dy = (f(x, y + h) - f(x, y - h)) / (2 * h)
    return np.array([df_dx, df_dy])

# Example usage
def f(x, y):
    return x**4 + y**4+ 2*x*y

point = (2, 3)
grad = gradient_2d(f, point[0], point[1])
print(f"Gradient at {point}: [{grad[0]:.6f}, {grad[1]:.6f}]")
print(f"Gradient magnitude: {np.linalg.norm(grad):.6f}")
```

```python
import numpy as np

def gradient_2d(f, x, y, h=1e-5):
    """
    Calculate gradient vector [∂f/∂x, ∂f/∂y] at point (x, y)
    Returns: numpy array
    """
    df_dx = (f(x + h, y,z) - f(x - h, y,z)) / (2 * h)
    df_dy = (f(x, y + h,z) - f(x, y - h,z)) / (2 * h)
    df_dy = (f(x, y + h,z) - f(x, y - h,z)) / (2 * h)
    return np.array([df_dx, df_dy])

# Example usage
def f(x, y):
    return x**4 + y**4+ 2*x*y

point = (2, 3)
grad = gradient_2d(f, point[0], point[1])
print(f"Gradient at {point}: [{grad[0]:.6f}, {grad[1]:.6f}]")
print(f"Gradient magnitude: {np.linalg.norm(grad):.6f}")
```

```python
import numpy as np

def gradient_3d(f, x, y, z, h=1e-1):
    """
    Calculate gradient vector [∂f/∂x, ∂f/∂y, ∂f/∂z] at point (x, y, z)
    Returns: numpy array
    """
    df_dx = (f(x + h, y, z) - f(x - h, y, z)) / (2 * h)
    df_dy = (f(x, y + h, z) - f(x, y - h, z)) / (2 * h)
    df_dz = (f(x, y, z + h) - f(x, y, z - h)) / (2 * h)
    return np.array([df_dx, df_dy, df_dz])

# Example usage
def f(x, y, z):
    return x**3 + y**3 + z**3

point = (1, 2, 3)
grad = gradient_3d(f, point[0], point[1], point[2])
print(f"Gradient at {point}: [{grad[0]:.6f}, {grad[1]:.6f}, {grad[2]:.6f}]")
print(f"Gradient magnitude: {np.linalg.norm(grad):.6f}")
```

```python
import numpy as np

def gradient_2d(f, x, y, h=1e-5):
    df_dx = (f(x + h, y) - f(x - h, y)) / (2 * h)
    df_dy = (f(x, y + h) - f(x, y - h)) / (2 * h)
    return np.array([df_dx, df_dy])

# Function: f(x, y) = x^3 + y^3 (CUBIC function)
def cubic(x, y):
    return x**3 + y**3

# Calculate gradient at different points with different h values
points = [(1, 1), (2, 0), (-1, 2)]
h_values = [1, 0.1, 0.01, 0.001, 1e-5]

print("Function: f(x, y) = x³ + y³")
print("Analytical gradient: [3x², 3y²]\n")

for point in points:
    x, y = point
    grad_analytical = np.array([3*x**2, 3*y**2])

    print(f"\nPoint: ({x}, {y})")
    print(f"Analytical gradient: [{grad_analytical[0]:.6f}, {grad_analytical[1]:.6f}]")
    print(f"\nNumerical gradients with different h:")

    for h in h_values:
        grad_numerical = gradient_2d(cubic, x, y, h)
        error = np.linalg.norm(grad_numerical - grad_analytical)
        print(f"  h = {h:8.5f}: [{grad_numerical[0]:10.6f}, {grad_numerical[1]:10.6f}]  Error: {error:.6f}")
    print("-" * 70)
```

```python
import numpy as np

def gradient_2d(f, x, y, h=1e-1):
    df_dx = (f(x + h, y) - f(x - h, y)) / (2 * h)
    df_dy = (f(x, y + h) - f(x, y - h)) / (2 * h)
    return np.array([df_dx, df_dy])

# Function: f(x, y) = x^2 - y^2 (saddle point)
def saddle(x, y):
    return x**3 - y**3

# Calculate gradient at different points
points = [(1, 1), (2, 1), (0, 2), (0, 0)]

print("Function: f(x, y) = x^2 - y^2")
print("Analytical gradient: [2x, -2y]\n")

for point in points:
    x, y = point
    grad_numerical = gradient_2d(saddle, x, y)
    grad_analytical = np.array([3*x**2, -3*y**2])

    print(f"Point: ({x}, {y})")
    print(f"Numerical gradient: [{grad_numerical[0]:.6f}, {grad_numerical[1]:.6f}]")
    print(f"Analytical gradient: [{grad_analytical[0]:.6f}, {grad_analytical[1]:.6f}]")
    print(f"Function value: {saddle(x, y):.6f}")
    print()
```

```python
import numpy as np

def gradient_3d(f, x, y, z, h=1e-5):
    df_dx = (f(x + h, y, z) - f(x - h, y, z)) / (2 * h)
    df_dy = (f(x, y + h, z) - f(x, y - h, z)) / (2 * h)
    df_dz = (f(x, y, z + h) - f(x, y, z - h)) / (2 * h)
    return np.array([df_dx, df_dy, df_dz])

# Function: f(x, y, z) = x^2 + y^2 + z^2
def sphere(x, y, z):
    return x**2 + y**2 + z**2

# Calculate gradient at different points
points = [(1, 1, 1), (2, 0, 1), (0, 0, 0), (1, -1, 2)]

print("Function: f(x, y, z) = x^2 + y^2 + z^2")
print("Analytical gradient: [2x, 2y, 2z]\n")

for point in points:
    x, y, z = point
    grad_numerical = gradient_3d(sphere, x, y, z)
    grad_analytical = np.array([2*x, 2*y, 2*z])

    print(f"Point: ({x}, {y}, {z})")
    print(f"Numerical gradient: [{grad_numerical[0]:.6f}, {grad_numerical[1]:.6f}, {grad_numerical[2]:.6f}]")
    print(f"Analytical gradient: [{grad_analytical[0]:.6f}, {grad_analytical[1]:.6f}, {grad_analytical[2]:.6f}]")
    print(f"Function value: {sphere(x, y, z):.6f}")
    print()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

def gradient_2d(f, x, y, h=1e-5):
    df_dx = (f(x + h, y) - f(x - h, y)) / (2 * h)
    df_dy = (f(x, y + h) - f(x, y - h)) / (2 * h)
    return np.array([df_dx, df_dy])

# Define function: paraboloid
def f(x, y):
    return x**2 + y**2

# Create grid for surface plot
x = np.linspace(-3, 3, 50)
y = np.linspace(-3, 3, 50)
X, Y = np.meshgrid(x, y)
Z = f(X, Y)

# Points where we'll calculate gradients
points = [(-2, -2), (-1, 1), (1, -1), (2, 2), (0, 0)]

# Calculate gradients at these points
gradients = []
for px, py in points:
    grad = gradient_2d(f, px, py)
    gradients.append(grad)

# Create figure with two subplots
fig = plt.figure(figsize=(16, 6))

# Subplot 1: 3D surface with gradient vectors
ax1 = fig.add_subplot(121, projection='3d')
ax1.plot_surface(X, Y, Z, cmap='viridis', alpha=0.6)

# Plot gradient vectors at selected points
for (px, py), grad in zip(points, gradients):
    pz = f(px, py)
    # Scale gradient for visibility
    scale = 0.5
    ax1.quiver(px, py, pz, grad[0], grad[1], 0,
               color='red', arrow_length_ratio=0.3, linewidth=2)
    ax1.scatter([px], [py], [pz], color='red', s=50)

ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.set_zlabel('f(x, y)')
ax1.set_title('Gradient Vectors on Surface f(x,y) = x² + y²\n(Gradient points in steepest ascent direction)')

# Subplot 2: Contour plot with gradient vectors
ax2 = fig.add_subplot(122)
contour = ax2.contour(X, Y, Z, levels=15, cmap='viridis')
ax2.clabel(contour, inline=True, fontsize=8)

# Plot gradient vectors on contour plot
for (px, py), grad in zip(points, gradients):
    # Normalize gradient for better visualization
    grad_norm = grad / (np.linalg.norm(grad) + 1e-10)
    ax2.arrow(px, py, grad_norm[0]*0.5, grad_norm[1]*0.5,
              head_width=0.2, head_length=0.15, fc='red', ec='red', linewidth=2)
    ax2.plot(px, py, 'ro', markersize=8)

ax2.set_xlabel('X')
ax2.set_ylabel('Y')
ax2.set_title('Gradient Vectors on Contour Plot\n(Gradient is perpendicular to contour lines)')
ax2.grid(True, alpha=0.3)
ax2.axis('equal')

plt.tight_layout()
plt.show()

# Print gradient information
print("Geometric Meaning of Gradient:\n")
print("1. Gradient points in the direction of STEEPEST ASCENT")
print("2. Gradient is PERPENDICULAR to contour lines (level curves)")
print("3. Gradient magnitude indicates how steep the function is\n")

for i, ((px, py), grad) in enumerate(zip(points, gradients)):
    mag = np.linalg.norm(grad)
    print(f"Point {i+1}: ({px}, {py})")
    print(f"  Gradient: [{grad[0]:.4f}, {grad[1]:.4f}]")
    print(f"  Magnitude: {mag:.4f}")
    print(f"  Direction (unit vector): [{grad[0]/mag:.4f}, {grad[1]/mag:.4f}]")
    print()
```
