```python
import pandas as pd 
import matplotlib.pyplot as plt
data = {
    'Name': [
        'Alice', 'Bob', 'Charlie', 'David', 'Eve',
        'Frank', 'Grace', 'Hannah', 'Ian', 'Jack',
        'Kiran', 'Liam', 'Mona', 'Nina', 'Owen',
        'Paul', 'Quinn', 'Riya', 'Sam', 'Tina'
    ],
    
    'Age': [
        25, 30, 35, None, 28,
        40, 29, 31, 27, 33,
        26, 34, 30, None, 28,
        36, 32, 24, 38, 27
    ],
    
    'City': [
        'New York', 'London', 'Paris', 'New York', 'London',
        'Berlin', 'Paris', 'London', 'New York', 'Berlin',
        'Delhi', 'Mumbai', 'Delhi', 'Bangalore', 'Chennai',
        'Delhi', 'Mumbai', 'Delhi', 'Bangalore', 'Chennai'
    ],
    
    'Salary': [
        50000, 60000, 75000, 55000, 52000,
        65000, 58000, 62000, 54000, 70000,
        48000, 72000, 61000, 59000, 56000,
        68000, 63000, 51000, 74000, 60000
    ]
}

df = pd.DataFrame(data)

df['Age'] = df['Age'].fillna(df['Age'].mean())

print(df)

avg_salary = df.groupby('City')['Salary'].mean()
print(avg_salary)

older_than_28 = df[df['Age'] > 28]
print(older_than_28)


plt.bar(df["City"],df["Salary"])
plt.xlabel("City")
plt.ylabel("Salary")
plt.show()
avg_salary = df.groupby('City')['Salary'].mean()

plt.pie(avg_salary, labels=avg_salary.index, autopct='%1.1f%%')
plt.title("Average Salary by City")
plt.show()
```

```python
import csv

def display():
    with open('inventory.csv', 'r') as f:
        reader = csv.reader(f)
        for row in reader:
            print(row)

def search(item):
    with open('inventory.csv', 'r') as f:
        reader = csv.reader(f)
        next(reader)  
        
        for row in reader:
            if row[0].lower() == item.lower():
                print("Item Found:", row)
                return
        print("Item not found")
def add_item():
    print("Adding new item...")   
    with open('inventory.csv', 'a', newline='') as f:
        item = input("Enter item name: ")
        qty = input("Enter quantity: ")
        price = input("Enter price: ")
        writer = csv.writer(f)
        writer.writerow([item, qty, price])


print("Inventory Data:")
display()

print("\nSearch Result:")
search("Pen")

print("\nAdd New Item:")
add_item()

print("\nUpdated Inventory:")
display()
```
