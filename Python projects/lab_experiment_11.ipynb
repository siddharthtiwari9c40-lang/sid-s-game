```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

plt.figure()
plt.plot(x, y, marker='o')
plt.title("Line Plot Example")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.grid()
plt.show()


categories = ['A', 'B', 'C', 'D', 'E']
values = [5, 7, 3, 8, 6]

plt.figure()
plt.bar(categories, values)
plt.title("Bar Chart Example")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()

data = [12, 15, 14, 10, 18, 20, 21, 19, 17, 16, 15, 14]

plt.figure()
plt.hist(data, bins=5)
plt.title("Histogram Example")
plt.xlabel("Value Range")
plt.ylabel("Frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt

x1 = [1, 2, 3]
y1 = [2, 4, 1]

x2 = [1, 2, 3]
y2 = [4, 1, 3]

plt.plot(x1, y1, marker='o', label="Line 1")
plt.plot(x2, y2, marker='o', label="Line 2")


plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Graph of Two Intersecting Lines")

plt.legend()

plt.grid()

plt.show()
```
