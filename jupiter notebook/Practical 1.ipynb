```python
# Program 1: Evaluating limit of (x^2 - 4)/(x - 2) as x approaches 2

import numpy as np

# Define the function
def f(x):
    return (x**2 - 4) / (x - 2)

# Point where we're evaluating the limit
a = 2.0

# Step sizes (approaching from left and right)
h_values = [0.1, 0.01, 0.001, 0.0001, 0.00001]

print("Evaluating limit of f(x) = (x^2 - 4)/(x - 2) as x approaches 2")
print("="*60)
print(f"{'h':>10} {'x (left)':>15} {'f(x) left':>15} {'x (right)':>15} {'f(x) right':>15}")
print("="*60)

for h in h_values:
    x_left = a - h      # Approach from left
    x_right = a + h     # Approach from right

    f_left = f(x_left)
    f_right = f(x_right)

    print(f"{h:>10} {x_left:>15.5f} {f_left:>15.5f} {x_right:>15.5f} {f_right:>15.5f}")

print("="*60)
print("As h → 0, both left and right limits approach 4")
print("Therefore, limit exists and equals 4")
```

```python
# Program 2: Evaluating limit of 1/x as x approaches 0

import numpy as np

def g(x):
    return 1 / x

a = 0.0
h_values = [0.1, 0.01, 0.001, 0.0001, 0.0000000001]

print("Evaluating limit of g(x) = 1/x as x approaches 0")
print("="*60)
print(f"{'h':>10} {'x (left)':>15} {'g(x) left':>15} {'x (right)':>15} {'g(x) right':>15}")
print("="*60)

for h in h_values:
    x_left = a - h
    x_right = a + h

    g_left = g(x_left)
    g_right = g(x_right)

    print(f"{h:>10} {x_left:>15.5f} {g_left:>15.5f} {x_right:>15.5f} {g_right:>15.5f}")

print("="*60)
print("Left limit → -∞, Right limit → +∞")
print("Since LHL ≠ RHL, the limit does NOT exist")
```

```python
# Program 3: Visualizing f(x) = (x^2 - 4)/(x - 2) with removable discontinuity at x=2

import numpy as np
import matplotlib.pyplot as plt

# Define the function
def f(x):
    # To avoid division by zero, we'll handle x=2 separately
    if abs(x - 2.0) < 0.0001:  # Very close to 2
        return None  # Don't calculate
    return (x**2 - 4) / (x - 2)

# Create x values, avoiding exactly x=2
x_left = np.linspace(-2, 1.99, 100)   # Left side: -2 to 1.99
x_right = np.linspace(2.01, 6, 100)   # Right side: 2.01 to 6

# Calculate y values
y_left = (x_left**2 - 4) / (x_left - 2)
y_right = (x_right**2 - 4) / (x_right - 2)

# Create the plot
plt.figure(figsize=(8, 6))

# Plot left and right parts
plt.plot(x_left, y_left, 'b-', linewidth=2, label='f(x)')
plt.plot(x_right, y_right, 'b-', linewidth=2)

# Mark the discontinuity point
plt.plot(2, 4, 'ro', markersize=10, markerfacecolor='white',
         markeredgewidth=2, label='Removable discontinuity at x=2')

# Mark where the limit approaches
plt.plot(2, 4, 'go', markersize=5, label='Limit = 4')

# Formatting
plt.xlabel('x', fontsize=12)
plt.ylabel('f(x)', fontsize=12)
plt.title('Removable Discontinuity: f(x) = (x²-4)/(x-2)', fontsize=14)
plt.grid(True, alpha=0.3)
plt.axhline(y=0, color='k', linewidth=0.5)
plt.axvline(x=0, color='k', linewidth=0.5)
plt.axvline(x=2, color='r', linewidth=0.5, linestyle='--', alpha=0.5)
plt.legend()
plt.ylim(-2, 8)
plt.show()

print("OBSERVATION:")
print("- The function has a 'hole' at x = 2")
print("- Left limit = Right limit = 4")
print("- But f(2) is undefined (division by zero)")
print("- This is a REMOVABLE discontinuity")
```

```python
# Program 4: Jump Discontinuity - Piecewise function

import numpy as np
import matplotlib.pyplot as plt

# Piecewise function:
# f(x) = x + 1,  if x < 2
# f(x) = x + 3,  if x >= 2

def f_jump(x):
    if x < 2:
        return x + 1
    else:
        return x + 3

# Create x values
x_left = np.linspace(-2, 2, 100)      # For x < 2
x_right = np.linspace(2, 6, 100)      # For x >= 2

# Calculate y values
y_left = x_left + 1
y_right = x_right + 3

# Create the plot
plt.figure(figsize=(8, 6))

# Plot both pieces
plt.plot(x_left, y_left, 'b-', linewidth=2, label='f(x) = x+1 (x<2)')
plt.plot(x_right, y_right, 'r-', linewidth=2, label='f(x) = x+3 (x≥2)')

# Mark the jump
plt.plot(2, 3, 'bo', markersize=8, markerfacecolor='white',
         markeredgewidth=2, label='Left limit = 3')
plt.plot(2, 5, 'ro', markersize=8, label='Right limit = 5')

# Formatting
plt.xlabel('x', fontsize=12)
plt.ylabel('f(x)', fontsize=12)
plt.title('Jump Discontinuity at x = 2', fontsize=14)
plt.grid(True, alpha=0.3)
plt.axhline(y=0, color='k', linewidth=0.5)
plt.axvline(x=0, color='k', linewidth=0.5)
plt.axvline(x=2, color='g', linewidth=0.5, linestyle='--', alpha=0.5)
plt.legend()
plt.ylim(-1, 9)
plt.show()

print("OBSERVATION:")
print("- Left limit as x → 2⁻ = 3")
print("- Right limit as x → 2⁺ = 5")
print("- Left limit ≠ Right limit")
print("- This is a JUMP discontinuity")
print("- The function 'jumps' from 3 to 5 at x = 2")
```

```python
# Program 5: Infinite Discontinuity - f(x) = 1/x at x=0

import numpy as np
import matplotlib.pyplot as plt

# Create x values, avoiding x=0
x_left = np.linspace(-5, -0.01, 200)    # Left side
x_right = np.linspace(0.01, 5, 200)     # Right side

# Calculate y values
y_left = 1 / x_left
y_right = 1 / x_right

# Create the plot
plt.figure(figsize=(8, 6))

# Plot both sides
plt.plot(x_left, y_left, 'b-', linewidth=2, label='f(x) = 1/x')
plt.plot(x_right, y_right, 'b-', linewidth=2)

# Formatting
plt.xlabel('x', fontsize=12)
plt.ylabel('f(x)', fontsize=12)
plt.title('Infinite Discontinuity: f(x) = 1/x at x = 0', fontsize=14)
plt.grid(True, alpha=0.3)
plt.axhline(y=0, color='k', linewidth=0.5)
plt.axvline(x=0, color='r', linewidth=2, linestyle='--',
            alpha=0.7, label='Vertical asymptote at x=0')
plt.legend()
plt.ylim(-10, 10)
plt.xlim(-5, 5)
plt.show()

print("OBSERVATION:")
print("- As x → 0⁻ (from left), f(x) → -∞")
print("- As x → 0⁺ (from right), f(x) → +∞")
print("- The function goes to infinity on both sides")
print("- This is an INFINITE discontinuity")
print("- x = 0 is a vertical asymptote")
```
